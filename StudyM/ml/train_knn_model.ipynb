```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder
from sklearn.neighbors import NearestNeighbors
from sklearn.model_selection import train_test_split

import pickle
```

```python
df = pd.read_csv("../dataset/study_material_dataset.csv")
df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
le_subject = LabelEncoder()
le_user_level = LabelEncoder()
le_difficulty = LabelEncoder()
le_topic = LabelEncoder()
```

```python
df['subject_enc'] = le_subject.fit_transform(df['subject'])
df['user_level_enc'] = le_user_level.fit_transform(df['user_level'])
df['difficulty_enc'] = le_difficulty.fit_transform(df['difficulty'])
df['topic_enc'] = le_topic.fit_transform(df['topic'])
```

```python
df[['subject', 'subject_enc',
'user_level', 'user_level_enc',
'difficulty', 'difficulty_enc',
'topic', 'topic_enc']].head()
```

```python
X = df[['subject_enc', 'user_level_enc', 'difficulty_enc', 'topic_enc']]
```

```python
X_train, X_test = train_test_split(
X,
test_size=0.2,
random_state=42
)
```

```python
knn = NearestNeighbors(
n_neighbors=5,
metric='euclidean'
)
```

```python
knn.fit(X_train)
```

```python
X_train
```

```python
X_test
```

```python
sample_input = np.array([[
le_subject.transform(['Python'])[0],
le_user_level.transform(['Beginner'])[0],
le_difficulty.transform(['Easy'])[0],
le_topic.transform(['Python Basics'])[0]
]])

distances, indices = knn.kneighbors(sample_input)

df.iloc[indices[0]][[
'subject',
'user_level',
'difficulty',
'topic',
'youtube_link',
'tutorial_site_link'
]]
```

```python
knn.fit(X)
```

```python
model_data = {
'knn_model': knn,
'dataset': df,
'subject_encoder': le_subject,
'user_level_encoder': le_user_level,
'difficulty_encoder': le_difficulty,
'topic_encoder': le_topic
}

with open("knn_model.pkl", "wb") as file:
    pickle.dump(model_data, file)
```
